```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
xs = np.arange(-1, 5, dtype=float)
ys = np.array([-4.0, 1.0, 6.0, 11.0, 16.0, 21.0], dtype=float)
print(f'xs = {xs}, ys = {ys}')
```

```python
regresion_lineal = np.polyfit(xs, ys, 1)
# y = b0x + b1
# beta --> b0 y alfa --> b1
beta, alfa = regresion_lineal
print(f'y = {round(beta, 1)}x + {round(alfa, 1)}')
```

```python
plt.plot(xs, ys, 'o', color='green', label='Datos')
plt.plot(xs, beta*xs + alfa, color='blue', label='Regresión Lineal')
plt.axhline(y = 0, color = 'red')
plt.axvline(x = 0, color = 'red')
plt.title('Say hello to the "Hello, World" of ML')
plt.xlabel('xs')
plt.ylabel('ys')
plt.legend()
plt.grid()
plt.show()
```

```python
# Perceptrón simple
capa = tf.keras.layers.Dense(units=1)
modelo = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),
    capa
  ])
modelo.compile(optimizer="sgd", loss="mse")
```

```python
print('Comenzando entrenamiento...')
historial = modelo.fit(xs, ys, epochs=1000, verbose=True)
print('Entrenamiento finalizado')
```

```python
valor = float(input('Ingrese un valor de x para predecir y: '))
prediccion = modelo.predict(np.array([[valor]]))
print(f'El valor de y para x = {valor} es: {prediccion}')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
xs = np.array([-40.0, -10.0, 0.0, 8.0, 15.0, 22.0, 38.0], dtype=float)
ys = np.array([-40.0, 14.0, 32.0, 46.0, 59.0, 72.0, 100.0], dtype=float)
print(f'xs = {xs}, ys = {ys}')
```

```python
# Solución analítica exacta: y = 1.8*x + 32
beta, alfa = np.polyfit(xs, ys, 1)
print(f'y = x * {round(beta, 1)} + {round(alfa, 1)}')
```

```python
plt.plot(xs, ys, 'o', color='green', label='Datos')
plt.plot(xs, beta*xs + alfa, color='blue', label='Regresión Lineal')
plt.axhline(y = 0, color = 'red')
plt.axvline(x = 0, color = 'red')
plt.title('Conversión Celsius (C°) a Fahrenheit (F°)')
plt.xlabel('xs')
plt.ylabel('ys')
plt.legend()
plt.grid()
plt.show()
```

```python
# Perceptrón simple
modelo = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1])
])
modelo.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.1), loss="mse")
```

```python
print('Comenzando entrenamiento...')
historial = modelo.fit(xs, ys, epochs=1000, verbose=0)
print('Entrenamiento finalizado')
```

```python
valor = float(input('Ingrese un valor de C° para predecir F°: '))
prediccion = modelo.predict(np.array([[valor]]), verbose=0)[0][0]
prediccion_real = valor * 9/5 + 32
print(f'Perceptrón predice: {prediccion:.2f}°F')
print(f'Fórmula real:       {prediccion_real:.2f}°F')
print(f'Error:              {abs(prediccion - prediccion_real):.4f}°F')
```
